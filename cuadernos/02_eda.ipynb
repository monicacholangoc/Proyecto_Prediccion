### Insight 1 · Sesgo de calificación

Las calificaciones altas concentran una proporción importante del dataset. Esto importa porque la utilidad no puede depender solo de estrellas, y obliga a usar métricas más robustas que accuracy.

### Insight 2 · Longitud del texto

Las reseñas útiles tienden a incluir más detalle. Esto sugiere que `review_len` puede aportar poder predictivo real.

### Insight 3 · Sentimiento

El sentimiento aporta señal, pero no reemplaza a la estructura textual. Una reseña muy positiva puede ser poco útil si no entrega contexto.

### Insight 4 · Incoherencia

Cuando el tono del texto no coincide con la calificación, la utilidad esperada se reduce. Esto respalda la creación de la variable `incoherente`.

### Insight 5 · Relación con el modelado
